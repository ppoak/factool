```python
import os
os.chdir('/home/wiikai/factor')

import factorlab as lab
import pandas as pd
import numpy as np
import quool

filter = quool.Factor("./data/filter-mask", code_level="order_book_id", date_level="date")
```

```python
start = '20160101'
stop = '20240701'
name = '000985.XSHG' 

df = lab.quotes_day.read(["limit_up", "high", "low", "limit_down", "open", 'st', 'suspended'], start=start, stop=stop)
df["IS_LIMIT_UP"] = np.where(df["high"] == df["limit_up"], True, False)
df["IS_LIMIT_DOWN"] = np.where(df["low"] == df["limit_down"], True, False)
df["alongLimitUp"] = np.where(df["open"] == df["limit_up"], True, False)
df["alongLimitDown"] = np.where(df["open"] == df["limit_down"], True, False)
limit_up = df["IS_LIMIT_UP"].unstack('order_book_id')
limit_down = df["IS_LIMIT_DOWN"].unstack('order_book_id')
alongLimitUp = df["alongLimitUp"].unstack('order_book_id')
alongLimitDown = df["alongLimitDown"].unstack('order_book_id')
st = df["st"].unstack('order_book_id')
suspended = df["suspended"].unstack('order_book_id')
universe = lab.index_weights.read(name ,start=start, stop=stop).isna()
nonrealizable =  st | suspended | limit_up | limit_down | alongLimitUp | alongLimitDown | universe
nonrealizable = nonrealizable.astype(bool)
```

```python
start = '20160101'
stop = '20240701'
name = 'cn'

df = lab.quotes_day.read(["limit_up", "high", "low", "limit_down", "open", 'st', 'suspended'], start=start, stop=stop)
df["IS_LIMIT_UP"] = np.where(df["high"] == df["limit_up"], True, False)
df["IS_LIMIT_DOWN"] = np.where(df["low"] == df["limit_down"], True, False)
df["alongLimitUp"] = np.where(df["open"] == df["limit_up"], True, False)
df["alongLimitDown"] = np.where(df["open"] == df["limit_down"], True, False)
limit_up = df["IS_LIMIT_UP"].unstack('order_book_id')
limit_down = df["IS_LIMIT_DOWN"].unstack('order_book_id')
alongLimitUp = df["alongLimitUp"].unstack('order_book_id')
alongLimitDown = df["alongLimitDown"].unstack('order_book_id')
st = df["st"].unstack('order_book_id')
suspended = df["suspended"].unstack('order_book_id')
nonrealizable = st | suspended | limit_up | limit_down | alongLimitUp | alongLimitDown
nonrealizable = nonrealizable.astype(bool)
nonrealizable
```

```python
nonrealizable.index.name = 'date'
nonrealizable = nonrealizable.sort_index(ascending=False).unstack()
nonrealizable.name = name

if name not in filter.columns:
    filter.add({name: nonrealizable.dtype})
filter.update(nonrealizable)
```
